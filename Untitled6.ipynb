```python
import pandas as pd
import numpy as np
def prepare_features(df):
    df = df.copy()
    dates = pd.to_datetime(df["publish_date"])
    df["publish_year"] = dates.dt.year
    df["publish_month"] = dates.dt.month
    df["publish_day"] = dates.dt.day
    df["publish_dayofyear"] = dates.dt.dayofyear
    df = df.drop(columns=["id", "publish_date"])
    return df

X = prepare_features(train.drop(columns=["target"]))
y = train["target"].astype(int)
X_test = prepare_features(test)
print("X ölçüsü:", X.shape)
print("y ölçüsü:", y.shape)
print("Test xüsusiyyətlərinin ölçüsü:", X_test.shape)
print("\nXüsusiyyətlərin ilk 5 sətri:")
display(X.head())
```

```python
dates = pd.to_datetime(train["publish_date"])
unique_dates = dates.drop_duplicates().sort_values().reset_index(drop=True)
cutoff_date = unique_dates.iloc[int(len(unique_dates) * 0.8)]
train_mask = dates < cutoff_date
valid_mask = dates >= cutoff_date
X_train = X.loc[train_mask]
X_valid = X.loc[valid_mask]
y_train = y.loc[train_mask]
y_valid = y.loc[valid_mask]
print("\nModelin öyrənəcəyi məlumatlar:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("Tarix aralığı:", dates[train_mask].min(), "-", dates[train_mask].max())
print("\nModelin yoxlanacağı məlumatlar:")
print("X_valid:", X_valid.shape)
print("y_valid:", y_valid.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
categorical_features = ["weekday", "channel"]
numeric_features = [col for col in X.columns if col not in categorical_features]
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])
categorical_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),("encoder", OneHotEncoder(handle_unknown="ignore"))])
preprocessor = ColumnTransformer([("numeric", numeric_pipeline, numeric_features),("categorical", categorical_pipeline, categorical_features)])
logistic_model = Pipeline([("preprocessor", preprocessor),("model", LogisticRegression(max_iter=2000))])
logistic_model.fit(X_train, y_train)
print("Logistic Regression modeli")
```

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,log_loss,confusion_matrix,classification_report)
valid_proba = logistic_model.predict_proba(X_valid)[:, 1]
valid_pred = (valid_proba >= 0.5).astype(int)
print("Accuracy:", round(accuracy_score(y_valid, valid_pred), 4))
print("Precision:", round(precision_score(y_valid, valid_pred), 4))
print("Recall:", round(recall_score(y_valid, valid_pred), 4))
print("F1-score:", round(f1_score(y_valid, valid_pred), 4))
print("ROC-AUC:", round(roc_auc_score(y_valid, valid_proba), 4))
print("Log Loss:", round(log_loss(y_valid, valid_proba), 4))
print("\nConfusion Matrix:")
print(confusion_matrix(y_valid, valid_pred))
print("\nClassification Report:")
print(classification_report(y_valid, valid_pred))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
def make_tree_preprocessor():
    return ColumnTransformer([("numeric",SimpleImputer(strategy="median"),numeric_features),("categorical",
         Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("encoder",OneHotEncoder(handle_unknown="ignore"))]),categorical_features)])
configs = {"GB_baseline": {"n_estimators": 150,"learning_rate": 0.05,"max_depth": 3,"min_samples_leaf": 20,"subsample": 0.8},
           "GB_more_trees": {"n_estimators": 250,"learning_rate": 0.03,"max_depth": 3,"min_samples_leaf": 20,"subsample": 0.8},
           "GB_shallow": {"n_estimators": 200,"learning_rate": 0.05,"max_depth": 2,"min_samples_leaf": 20,"subsample": 0.8},
           "GB_regularized": {"n_estimators": 250,"learning_rate": 0.03,"max_depth": 2,"min_samples_leaf": 30,"subsample": 0.9},
           "GB_faster_learning": {"n_estimators": 150,"learning_rate": 0.08,"max_depth": 2,"min_samples_leaf": 30,"subsample": 0.8}}
tuned_models = {}
tuned_results = []
for name, params in configs.items():
    print(f"{name} öyrədilir...")
    model = Pipeline([("preprocessor", make_tree_preprocessor()),("model", GradientBoostingClassifier(**params,random_state=42))])
    model.fit(X_train, y_train)
    probabilities = model.predict_proba(X_valid)[:, 1]
    predictions = (probabilities >= 0.5).astype(int)
    tuned_results.append({"Model": name,"Accuracy": accuracy_score(y_valid, predictions),"F1": f1_score(y_valid, predictions),
                          "ROC-AUC": roc_auc_score(y_valid, probabilities),"Log Loss": log_loss(y_valid, probabilities)})
    tuned_models[name] = model
tuning_table = pd.DataFrame(tuned_results)
tuning_table = tuning_table.sort_values("ROC-AUC", ascending=False).reset_index(drop=True)
display(tuning_table.round(4))
```

```python
import numpy as np
import pandas as pd
def engineer_features(df):
    df = df.copy()
    words = df["n_tokens_content"] + 1
    df["title_body_ratio"] = df["n_tokens_title"] / words
    df["links_per_100_words"] = df["num_hrefs"] * 100 / words
    df["images_per_100_words"] = df["num_imgs"] * 100 / words
    df["videos_per_100_words"] = df["num_videos"] * 100 / words
    df["self_href_ratio"] = (df["num_self_hrefs"] / (df["num_hrefs"] + 1))
    df["total_media"] = df["num_imgs"] + df["num_videos"]
    df["positive_negative_gap"] = (df["avg_positive_polarity"]+ df["avg_negative_polarity"])
    df["positive_polarity_range"] = (df["max_positive_polarity"]- df["min_positive_polarity"])
    df["negative_polarity_range"] = (df["max_negative_polarity"]- df["min_negative_polarity"])
    df["reference_avg_max_ratio"] = (df["self_reference_avg_shares"]/ (df["self_reference_max_shares"] + 1))
    df["month_sin"] = np.sin(2 * np.pi * df["publish_month"] / 12)
    df["month_cos"] = np.cos(2 * np.pi * df["publish_month"] / 12)
    df["dayyear_sin"] = np.sin(2 * np.pi * df["publish_dayofyear"] / 365.25)
    df["dayyear_cos"] = np.cos(2 * np.pi * df["publish_dayofyear"] / 365.25)
    return df
X_eng = engineer_features(X)
X_test_eng = engineer_features(X_test)
print("Əvvəlki xüsusiyyət sayı:", X.shape[1])
print("Yeni xüsusiyyət sayı:", X_eng.shape[1])
display(X_eng.head())
```

```python
import xgboost as xgb
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import roc_auc_score, log_loss
cat_cols = ["weekday", "channel"]
num_cols = [col for col in X_train.columns if col not in cat_cols]
num_prep = Pipeline([("imputer", SimpleImputer(strategy="median"))])
cat_prep = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),("encoder", OneHotEncoder(handle_unknown="ignore"))])
xgb_preprocessor = ColumnTransformer([("num", num_prep, num_cols),("cat", cat_prep, cat_cols)])
xgb_model = Pipeline([("preprocessor", xgb_preprocessor),("model", xgb.XGBClassifier(n_estimators=800,max_depth=4,learning_rate=0.03,min_child_weight=5,
                                                                                     subsample=0.85,colsample_bytree=0.85,reg_alpha=0.1,reg_lambda=5,
                                                                                     objective="binary:logistic",eval_metric="auc",tree_method="hist",
                                                                                     random_state=42,n_jobs=-1))])

xgb_model.fit(X_train, y_train)
xgb_proba = xgb_model.predict_proba(X_valid)[:, 1]
print("XGBoost ROC-AUC:",round(roc_auc_score(y_valid, xgb_proba), 5))
print("XGBoost Log Loss:",round(log_loss(y_valid, xgb_proba), 5))
```
